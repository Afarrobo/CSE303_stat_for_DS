```python
# Step 1: Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Step 2: Load dataset
df = pd.read_csv('/kaggle/input/loan-data-0111/LoanData_Preprocessed_v1.1.csv')

# Step 3: Basic info
print("Shape of dataset:", df.shape)
print("Column info:\n")
print(df.info())
print("\nFirst 5 rows:\n", df.head())
```

```python
df.dtypes
```

```python
  # Shows both numeric and categorical summaries

print("Columns:", df.columns.tolist())
df.describe(include='all')  
```

```python

# Describe Numerical Columns
num_col = 'debtinc'

mean_val = df[num_col].mean()
median_val = df[num_col].median()
var_val = df[num_col].var()
std_val = df[num_col].std()

print(f"{num_col} - Mean: {mean_val}, Median: {median_val}, Variance: {var_val}, Std Dev: {std_val}")
```

```python
# Describe Numerical Columns 2 
num_col2 = 'creddebt'

mean_val2 = df[num_col2].mean()
median_val2 = df[num_col2].median()
var_val2 = df[num_col2].var()
std_val2 = df[num_col2].std()

print(f"{num_col2} - Mean: {mean_val2}, Median: {median_val2}, Variance: {var_val2}, Std Dev: {std_val2}")
```

```python
# Describe Numerical Columns 3 
num_col23 = 'othdebt'

mean_val23 = df[num_col23].mean()
median_val23 = df[num_col23].median()
var_val23 = df[num_col23].var()
std_val23 = df[num_col23].std()

print(f"{num_col23} - Mean: {mean_val23}, Median: {median_val23}, Variance: {var_val23}, Std Dev: {std_val23}")
```


# Catagorical or discrete column is age , employ , adress, income , ed . So , count the frequency and percentage of each type of values.


```python
# age's all

cat_col = 'age'
value_counts = df[cat_col].value_counts()
percent = df[cat_col].value_counts(normalize=True) * 100

print(value_counts)
print(percent)
```


# **Barchart comparing 3 column . All of them are  Catagorical columns# ******


```python
# List of categorical columns to compare
columns = ['age', 'employ', 'income']

# Create subplot
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Loop through each column and create bar plot
for i, col in enumerate(columns):
    counts = df[col].value_counts()
    axes[i].bar(counts.index, counts.values, color='skyblue')
    axes[i].set_title(f'{col} Distribution')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Count')
    axes[i].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

# Histogram for Continuous numerical data

```python
plt.figure(figsize=(8, 5))
plt.hist(df['debtinc'].dropna(), bins=20, color='skyblue', edgecolor='black')

# Customized titles and labels
plt.title("Distribution of Debt-to-Income Ratio", fontsize=14, fontweight='bold')
plt.xlabel("Debt-to-Income Ratio (%)", fontsize=12)
plt.ylabel("Number of Applicants", fontsize=12)

# Grid and style
plt.grid(True, linestyle='--', alpha=0.9)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df['creddebt'].dropna(), bins=20, color='lightgreen', edgecolor='black')

# Customized titles and labels
plt.title("Distribution of Credit Debt", fontsize=14, fontweight='bold')
plt.xlabel("credit debt ", fontsize=12)
plt.ylabel("No. of Applicants", fontsize=12)

# Grid and style
plt.grid(True, linestyle='--', alpha=0.9)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.tight_layout()

plt.show()
```

```python
total_creddebt1 = df['creddebt'].sum()
total_othdebt1 = df['othdebt'].sum()

plt.pie([total_creddebt1, total_othdebt1],
        labels=['Credit Debt', 'Other Debt'],
        autopct='%1.1f%%',
        colors=['orange', 'lightblue'],
        startangle=140)

plt.title('Total Debt Composition')
plt.axis('equal')
plt.show()
```

```python
# Boxplot: income distribution by education
sns.boxplot(x='ed', y='income', data=df, palette='Set3')
plt.title('Income Distribution by Education Level')
plt.xlabel('Education Level')
plt.ylabel('Income (in thousands)')
plt.tight_layout()
plt.show()
```

#  Correlation heat-map with each column. 

```python
# Select only numerical columns for correlation
numerical_df = df.select_dtypes(include=['int64', 'float64'])

# Calculate correlation matrix
corr_matrix = numerical_df.corr()

# Plot the heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap of Numerical Features", fontsize=14, fontweight='bold')
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

# Check Missing Values for csv file-1:

```python
# Check missing values in each column
missing_values = df.isnull().sum()
print("Missing values per column:\n", missing_values)
```

```python
# Fill missing numeric values with median (better for skewed data)
df['age'] = df['age'].fillna(df['age'].median())
df['income'] = df['income'].fillna(df['income'].median())
df['ed'] = df['ed'].fillna(df['ed'].median())
```

**We use Median because it is not affected by outliers**

```python
# Check for missing values in the entire DataFrame
missing_values = df.isnull().sum()

# Display columns with missing values
print(missing_values[missing_values > 0])
```
